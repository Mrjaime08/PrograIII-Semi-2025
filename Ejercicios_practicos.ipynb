```python
def suma(num1, num2):
    return num1 + num2

def resta(num1, num2):
    return num1 - num2

def multiplicar(num1, num2):
    return num1 * num2

def dividir(num1, num2):
    if num2 != 0:
        return num1 / num2
    else:
        return "No se puede dividir entre cero"

def mitad(num):
    return num / 2

# Menú de operaciones
print("Selecciona una operación:")
print("1. Suma")
print("2. Resta")
print("3. Multiplicación")
print("4. División")
print("5. Mitad de un número")

opcion = input("Opción (1-5): ")

# Operaciones que requieren dos números
if opcion in ['1', '2', '3', '4']:
    num1 = float(input("Num1: "))
    num2 = float(input("Num2: "))

    if opcion == '1':
        resultado = suma(num1, num2)
    elif opcion == '2':
        resultado = resta(num1, num2)
    elif opcion == '3':
        resultado = multiplicar(num1, num2)
    elif opcion == '4':
        resultado = dividir(num1, num2)

# Operación que requiere un solo número
elif opcion == '5':
    num = float(input("Ingresa un número: "))
    resultado = mitad(num)
else:
    resultado = "Opción inválida"

print("Resultado:", resultado)
```

```python
# Diccionarios base
tasas_conversion = {
    "USD": 1.0, "EUR": 0.85, "GBP": 0.74, "CHF": 0.91,
    "JPY": 110.27, "HKD": 7.77, "CAD": 1.25, "CNY": 6.47, "COL": 8.75, "MXN": 16.9
}

longitud = {"metro": 1, "cm": 0.01, "mm": 0.001, "km": 1000, "pulgada": 0.0254, "pie": 0.3048, "yarda": 0.9144, "milla": 1609.34, "nanometro": 1e-9, "micrometro": 1e-6}
masa = {"kg": 1, "g": 0.001, "mg": 1e-6, "lb": 0.4536, "oz": 0.0283, "tonelada": 1000, "quintal": 45.36, "stone": 6.35, "microgramo": 1e-9}
almacenamiento = {"byte": 1, "kb": 1024, "mb": 1024*2, "gb": 10243, "tb": 10244, "bit": 0.125, "nibble": 0.5, "pb": 10245, "eb": 1024*6}
tiempo = {"segundo": 1, "minuto": 60, "hora": 3600, "día": 86400, "semana": 604800, "mes": 2.63e6, "año": 3.15e7, "milisegundo": 0.001, "microsegundo": 1e-6}
volumen = {"litro": 1, "ml": 0.001, "galón": 3.785, "onza": 0.0296, "m3": 1000, "cm3": 0.001, "taza": 0.24, "pinta": 0.473, "cuarto": 0.946}
area = {"m2": 1, "cm2": 0.0001, "mm2": 1e-6, "km2": 1e6, "hectarea": 10000, "acre": 4046.86, "pie2": 0.0929, "pulgada2": 0.000645, "yarda2": 0.836, "manzana": 6987.4}
manzanas = {"manzana": 1, "m2": 1/6987.4, "hectarea": 1/0.6987, "acre": 1/1.726, "pie2": 1/75264, "km2": 1/143, "cm2": 1/69874000, "mm2": 1/698740000, "tarea": 1/6.98}
tareas = {"tarea": 1, "m2": 1/350, "manzana": 1/6.98, "hectarea": 1/3.5, "acre": 1/1.4, "pie2": 1/37674, "km2": 1/286, "cm2": 1/3500000, "mm2": 1/35000000}

# Funciones de conversión
def convertir_moneda(cantidad, origen, destino):
    if origen not in tasas_conversion or destino not in tasas_conversion:
        return None
    usd = cantidad / tasas_conversion[origen]
    return usd * tasas_conversion[destino]

def convertir_general(cantidad, origen, destino, diccionario):
    if origen not in diccionario or destino not in diccionario:
        return None
    base = cantidad * diccionario[origen]
    return base / diccionario[destino]

def conversor_personalizado(tipo, cantidad):
    if tipo == "caja":
        return f"{int(cantidad)} unidades = {int(cantidad) // 25} cajas y {int(cantidad) % 25} unidades"
    elif tipo == "saco":
        return f"{int(cantidad)} libras = {int(cantidad) // 50} sacos y {int(cantidad) % 50} libras"
    elif tipo == "fardo":
        return f"{int(cantidad)} unidades = {int(cantidad) // 5} fardos y {int(cantidad) % 5} unidades"
    else:
        return "Tipo no válido."

# Funciones adicionales de la guía
def calcular_deducciones(sueldo):
    isss = sueldo * 0.03
    afp = sueldo * 0.0625
    if sueldo <= 472.00:
        isr = 0
    elif sueldo <= 895.24:
        isr = (sueldo - 472.00) * 0.10 + 17.67
    elif sueldo <= 2038.10:
        isr = (sueldo - 895.24) * 0.20 + 60.00
    else:
        isr = (sueldo - 2038.10) * 0.30 + 288.57
    total = isss + afp + isr
    neto = sueldo - total

    # Calculos adicionales de paga
    paga_mensual = neto
    paga_anual = neto * 12
    paga_quincenal = neto / 2
    # Assuming 4 weeks in a month for simplicity, and 5 working days per week
    paga_semanal = neto / 4

    # New calculations for daily and hourly pay
    # Assuming 22 working days in a month (approx. 4.4 weeks * 5 days/week)
    paga_diaria = neto / 22
    # Assuming 8 working hours per day
    paga_hora = paga_diaria / 8

    return total, neto, paga_semanal, paga_quincenal, paga_mensual, paga_anual, paga_diaria, paga_hora

def calcular_agua(metros):
    if metros <= 18:
        return 6.00
    elif metros <= 28:
        return 6.00 + (metros - 18) * 0.45
    else:
        return 6.00 + 10 * 0.45 + (metros - 28) * 0.65

def tabla_meses(n):
    print("\nMes\tAcumulado")
    for mes in range(1, n+1):
        acumulado = mes * (mes + 1) // 2
        print(f"{mes}\t{acumulado}")

# Menú principal
def menu():
    print("\n--- Menú de opciones ---")
    print("1. Conversor de monedas")
    print("2. Conversores físicos")
    print("3. Conversores personalizados")
    print("4. Deducciones sobre sueldo")
    print("5. Consumo de agua potable")
    print("6. Tabla de meses acumulados")
    return input("Seleccione una opción (1-6): ")

# Ejecución principal
while True:
    opcion = menu()
    if opcion == "1":
        print("Monedas disponibles:", ", ".join(tasas_conversion.keys()))
        o = input("Moneda origen: ").upper()
        # Check origin currency FIRST
        if o not in tasas_conversion:
            print("⚠️ Moneda de origen no disponible. Por favor, verifique.")
        else:
            d = input("Moneda destino: ").upper()
            # If origin is valid, then check destination
            if d not in tasas_conversion:
                print("⚠️ Moneda de destino no disponible. Por favor, verifique.")
            else:
                try:
                    c = float(input("Cantidad: "))
                    r = convertir_moneda(c, o, d)
                    print(f"{c} {o} = {r:.2f} {d}")
                except ValueError:
                    print("⚠️ Entrada inválida. Solo se aceptan números.")
    elif opcion == "2":
        grupos = [longitud, masa, almacenamiento, tiempo, volumen, area, manzanas, tareas]
        nombres = ["Longitud", "Masa", "Almacenamiento", "Tiempo", "Volumen", "Área", "Manzanas", "Tareas"]
        for i, nombre in enumerate(nombres, 1):
            print(f"{i}. {nombre}")
        try:
            sub = int(input("Seleccione tipo (1-8): "))
            dic = grupos[sub - 1]
            print("Unidades disponibles:", ", ".join(dic.keys()))
            o = input("Unidad origen: ").lower()
            d = input("Unidad destino: ").lower()
            if o not in dic: # Check origin unit first
                print("⚠️ Unidad de origen no disponible. Por favor, verifique.")
            elif d not in dic: # Check destination unit if origin is valid
                print("⚠️ Unidad de destino no disponible. Por favor, verifique.")
            else:
                c = float(input("Cantidad: "))
                r = convertir_general(c, o, d, dic)
                print(f"{c} {o} = {r:.4f} {d}")
        except (IndexError, ValueError): # Catch errors from int() conversion or list index
            print("⚠️ Entrada inválida o selección fuera de rango.")
    elif opcion == "3":
        tipo = input("Tipo (caja, saco, fardo): ").lower()
        if tipo not in ["caja", "saco", "fardo"]:
            print("⚠️ Tipo de conversión personalizado no válido.")
        else:
            try:
                cantidad = float(input("Cantidad: "))
                # Convert quantity to int for specific custom converters
                if tipo in ["caja", "saco", "fardo"]:
                    cantidad = int(cantidad) # Ensure integer for remainder operations
                print(conversor_personalizado(tipo, cantidad))
            except ValueError:
                print("⚠️ Entrada inválida. Solo se aceptan números.")
    elif opcion == "4":
        try:
            sueldo = float(input("Ingrese el sueldo bruto: $"))
            total, neto, semanal, quincenal, mensual, anual, diaria, hora = calcular_deducciones(sueldo)
            print(f"Total deducciones: ${total:.2f}")
            print(f"Sueldo neto: ${neto:.2f}")
            print("\n--- Desglose de Pagos Netos ---")
            print(f"Pago por hora (estimado): ${hora:.2f}")
            print(f"Pago diario (estimado): ${diaria:.2f}")
            print(f"Pago semanal (estimado): ${semanal:.2f}")
            print(f"Pago quincenal: ${quincenal:.2f}")
            print(f"Pago mensual: ${mensual:.2f}")
            print(f"Pago anual: ${anual:.2f}")
        except ValueError:
            print("⚠️ Entrada inválida. Solo se aceptan números.")
    elif opcion == "5":
        try:
            metros = float(input("Ingrese metros cúbicos consumidos: "))
            pago = calcular_agua(metros)
            print(f"Total a pagar: ${pago:.2f}")
        except ValueError:
            print("⚠️ Entrada inválida. Solo se aceptan números.")
    elif opcion == "6":
        try:
            n = int(input("¿Cuántos meses desea acumular?: "))
            if n < 1:
                print("⚠️ Por favor, ingrese un número positivo de meses.")
            else:
                tabla_meses(n)
        except ValueError:
            print("⚠️ Entrada inválida. Solo se aceptan números enteros.")
    else:
        print("⚠️ Opción inválida. Intente nuevamente.")

    if input("\n¿Desea realizar otra operación? (s/n): ").lower() != "s":
        print("Gracias por usar el sistema. ¡Hasta pronto!")
        break
```

**CALCULAR CONSUMO DE AGUA**

```python
def calcular_tarifa_agua(consumo_mts):
    """
    Calcula el valor a pagar por el consumo de agua potable
    basado en un sistema de tarifas por tramos.
    """

    # Definimos las constantes para que el código sea más legible
    CUOTA_FIJA = 6.00
    PRECIO_TRAMO_2 = 0.45
    PRECIO_TRAMO_3 = 0.65

    LIMITE_TRAMO_1 = 18
    LIMITE_TRAMO_2 = 28

    costo_total = 0.0

    # Condicional para determinar en qué tramo se encuentra el consumo
    if consumo_mts <= 0:
        costo_total = 0.0 # No hay costo si no hay consumo
    elif consumo_mts <= LIMITE_TRAMO_1:
        # TRAMO 1: Se paga únicamente la cuota fija
        costo_total = CUOTA_FIJA
    elif consumo_mts <= LIMITE_TRAMO_2:
        # TRAMO 2: Cuota fija + costo del exceso sobre el tramo 1
        exceso_tramo_1 = consumo_mts - LIMITE_TRAMO_1
        costo_total = CUOTA_FIJA + (exceso_tramo_1 * PRECIO_TRAMO_2)
    else: # El consumo es mayor a 28
        # TRAMO 3: Se suman los costos de todos los tramos

        # Costo del primer tramo (completo)
        costo_tramo_1 = CUOTA_FIJA

        # Costo del segundo tramo (completo: 10 metros del 19 al 28)
        metros_tramo_2 = LIMITE_TRAMO_2 - LIMITE_TRAMO_1
        costo_tramo_2 = metros_tramo_2 * PRECIO_TRAMO_2

        # Costo del exceso sobre el tramo 2
        exceso_tramo_2 = consumo_mts - LIMITE_TRAMO_2
        costo_tramo_3 = exceso_tramo_2 * PRECIO_TRAMO_3

        costo_total = costo_tramo_1 + costo_tramo_2 + costo_tramo_3

    return costo_total

# --- Bloque principal para ejecutar el programa ---
if __name__ == "__main__":
    try:
        # Pedimos al usuario que ingrese los datos
        metros_consumidos = float(input("Ingrese el número de metros cúbicos consumidos: "))

        # Llamamos a la función para hacer el cálculo
        valor_a_pagar = calcular_tarifa_agua(metros_consumidos)

        # Mostramos el resultado formateado a dos decimales, como es usual en dinero
        print(f"El consumo fue de {metros_consumidos} m³.")
        print(f"El valor total a pagar es: ${valor_a_pagar:.2f}")

    except ValueError:
        # Manejo de error si el usuario no ingresa un número
        print("Error: Por favor, ingrese un valor numérico válido.")
```

CALCULAR SUELDO + IMPUESTOS

```python
def calcular_deducciones(sueldo):

    isss = min(sueldo * 0.03,30.00)
    afp = sueldo * 0.0725

    sueldo_gravado = sueldo - (isss + afp)

    if sueldo_gravado <= 550.00:
        isr = 0
    elif sueldo_gravado <= 895.24:
        isr = (sueldo_gravado - 550.00) * 0.10 + 17.67
    elif sueldo_gravado <= 2038.10:
        isr = (sueldo_gravado - 895.24) * 0.20 + 60.00
    else:
        isr = (sueldo_gravado - 2038.10) * 0.30 + 288.57

    if sueldo <= 0:
        print("Error: Digite un Numero Positivo")
        return (0, 0, 0, 0, 0)

    total_deducciones = isss + afp + isr
    sueldo_neto = sueldo - total_deducciones

    return (isss, afp, isr, total_deducciones, sueldo_neto)
```

```python
sueldo = float(input("Ingrese su sueldo: "))
resultado = calcular_deducciones(sueldo)

print("\n--- RESULTADOS ---")
print(f"ISSS (3%): ${resultado[0]:.2f}")
print(f"AFP (7.25%): ${resultado[1]:.2f}")
print(f"ISR: ${resultado[2]:.2f}")
print(f"Total deducciones: ${resultado[3]:.2f}")
print(f"Sueldo neto: ${resultado[4]:.2f}")
```

```python
# --- Constantes de Conversión ---
# Unidades de Empaque/Inventario
UNIDADES_POR_CAJA = 25.0
LIBRAS_POR_SACO = 50.0
UNIDADES_POR_FARDO = 5.0

# --- NUEVAS CONSTANTES ---
# Longitud
PIES_POR_METRO = 3.28084

# Almacenamiento Digital
MEGABYTES_POR_GIGABYTE = 1024.0

# La conversión de temperatura es una fórmula, no una constante, así que se manejará directamente.

def mostrar_menu():
    """Imprime el menú de opciones ampliado para el usuario."""
    print("\n--- Conversor Universal ---")
    print("--- Empaque/Inventario ---")
    print("1. Cajas a Unidades")
    print("2. Unidades a Cajas")
    print("3. Sacos a Libras")
    print("4. Libras a Sacos")
    print("5. Fardos a Unidades")
    print("6. Unidades a Fardos")
    print("\n--- Longitud ---")
    print("7. Metros a Pies")
    print("8. Pies a Metros")
    print("\n--- Temperatura ---")
    print("9. Celsius a Fahrenheit")
    print("10. Fahrenheit a Celsius")
    print("\n--- Almacenamiento Digital ---")
    print("11. Gigabytes (GB) a Megabytes (MB)")
    print("12. Megabytes (MB) a Gigabytes (GB)")
    print("\n13. Salir")

# --- Bucle principal del programa ---
while True:
    mostrar_menu()
    opcion = input("Seleccione una opción: ")

    # Usamos un bloque try-except general para el input numérico
    try:
        # --- Conversiones de Empaque ---
        if opcion == '1':
            cajas = float(input("Ingrese la cantidad de cajas: "))
            unidades = cajas * UNIDADES_POR_CAJA
            print(f"-> {cajas} cajas son {unidades} unidades.")

        elif opcion == '2':
            unidades = float(input("Ingrese la cantidad de unidades: "))
            cajas = unidades / UNIDADES_POR_CAJA
            print(f"-> {unidades} unidades son {cajas:.2f} cajas.")

        elif opcion == '3':
            sacos = float(input("Ingrese la cantidad de sacos: "))
            libras = sacos * LIBRAS_POR_SACO
            print(f"-> {sacos} sacos son {libras} libras.")

        elif opcion == '4':
            libras = float(input("Ingrese la cantidad de libras: "))
            sacos = libras / LIBRAS_POR_SACO
            print(f"-> {libras} libras son {sacos:.2f} sacos.")

        elif opcion == '5':
            fardos = float(input("Ingrese la cantidad de fardos: "))
            unidades = fardos * UNIDADES_POR_FARDO
            print(f"-> {fardos} fardos son {unidades} unidades.")

        elif opcion == '6':
            unidades = float(input("Ingrese la cantidad de unidades: "))
            fardos = unidades / UNIDADES_POR_FARDO
            print(f"-> {unidades} unidades son {fardos:.2f} fardos.")

        # --- NUEVAS CONVERSIONES ---
        # --- Longitud ---
        elif opcion == '7':
            metros = float(input("Ingrese la cantidad de metros: "))
            pies = metros * PIES_POR_METRO
            print(f"-> {metros} metros son {pies:.2f} pies.")

        elif opcion == '8':
            pies = float(input("Ingrese la cantidad de pies: "))
            metros = pies / PIES_POR_METRO
            print(f"-> {pies} pies son {metros:.2f} metros.")

        # --- Temperatura (usando fórmulas) ---
        elif opcion == '9':
            celsius = float(input("Ingrese los grados Celsius: "))
            fahrenheit = (celsius * 9/5) + 32
            print(f"-> {celsius}°C son {fahrenheit:.2f}°F.")

        elif opcion == '10':
            fahrenheit = float(input("Ingrese los grados Fahrenheit: "))
            celsius = (fahrenheit - 32) * 5/9
            print(f"-> {fahrenheit}°F son {celsius:.2f}°C.")

        # --- Almacenamiento Digital ---
        elif opcion == '11':
            gb = float(input("Ingrese la cantidad de Gigabytes (GB): "))
            mb = gb * MEGABYTES_POR_GIGABYTE
            print(f"-> {gb} GB son {mb} MB.")

        elif opcion == '12':
            mb = float(input("Ingrese la cantidad de Megabytes (MB): "))
            gb = mb / MEGABYTES_POR_GIGABYTE
            print(f"-> {mb} MB son {gb:.2f} GB.")

        # --- Opción de Salida ---
        elif opcion == '13':
            print("¡Gracias por usar el conversor! Adiós.")
            break # Rompe el bucle while y termina el programa

        else:
            print("Opción no válida. Por favor, intente de nuevo.")

    except ValueError:
        print("Error: Entrada inválida. Por favor, ingrese solo números.")
```
