```python
#ENCONTRAR LA PERSONA DE MAYOR EDAD EN UNA LISTA DE DATOS
personas = [
  {"nombre": "Luis", "edad": 30},
  {"nombre": "Juan", "edad": 40},
  {"nombre": "Jorge", "edad": 20},
  {"nombre": "Jose", "edad": 10}
]

persona_mas_vieja = personas[0]

for persona in personas:
  if persona["edad"] > persona_mas_vieja["edad"]:
    persona_mas_vieja = persona

print(f"La persona de mayor edad es: {persona_mas_vieja['nombre']} con {persona_mas_vieja['edad']} años.")
```
